### Análises bidimensionais

Para começar as análises bidimensionais, contei quantas ocorrências ocorreram por região. É possível ver que a capital do Rio de Janeiro é bem mais violenta.

```python
alt.Chart(dfanos).mark_bar().encode(
    alt.X('regiao' , axis=alt.Axis(title='Região')),
    alt.Y('sum(quantidade_registro_ocorrencia)', axis=alt.Axis(title='Número total de ocorrências')),
    alt.Color('regiao:N',
        scale=alt.Scale(
        domain=['Baixada Fluminense', 'Capital', 'Grande Niterói', 'Interior'],
        range= ['#830A1E', '#8C0F2F', '#A61237', '#CF1B51']
    ))
).properties(
    title='Número de ocorrências por região'
)
```

Decidi contar também as ocorrências por região de posse e tráfico de drogas, achando que isso teria mais frequência na capital.

```python
alt.Chart(dfanos).mark_bar().encode(
    alt.X('regiao' , axis=alt.Axis(title='Região')),
    alt.Y('sum(quantidade_registro_posse_drogas)', axis=alt.Axis(title='Número total de registros de posse de droga')),
    alt.Color('regiao:N',
        scale=alt.Scale(
        domain=['Baixada Fluminense', 'Capital', 'Grande Niterói', 'Interior'],
        range= ['#830A1E', '#8C0F2F', '#A61237', '#CF1B51']
    ))
).properties(
    title='Registros de posse de droga por região'
)
```

```python
alt.Chart(dfanos).mark_bar().encode(
    alt.X('regiao' , axis=alt.Axis(title='Região')),
    alt.Y('sum(quantidade_registro_trafico_drogas)', axis=alt.Axis(title='Número total de registros de tráfico de droga')),
    alt.Color('regiao:N',
        scale=alt.Scale(
        domain=['Baixada Fluminense', 'Capital', 'Grande Niterói', 'Interior'],
        range= ['#830A1E', '#8C0F2F', '#A61237', '#CF1B51']
    ))
).properties(
    title='Registros de tráfico de droga por região'
)
```

Vi que o gráfico era muito concentrado no começo, então decidi torná-lo interativo para que pudessemos ver masi de perto o que acontecia

```python
alt.Chart(dfanos).mark_circle().encode(
    alt.X('quantidade_registro_posse_drogas', axis=alt.Axis(title='Quantidade de registros de posse de drogas')),
    alt.Y('quantidade_registro_trafico_drogas', axis=alt.Axis(title='Quantidade de registros de tráfico')),
    alt.Color('count()'),
    alt.Size('count()')
).properties(
    width= 1000,
    title='Relação entre tráfico e posse de drogas'
).interactive()
```

Por fim, decidi colocar no gráfico a opção de destacar as regiões para que seja possível analisar elas indiviualmente. Agora, fica evidente que o interior tem muito mais registros de tráfico e posse de drogas do que as demais regiões.

```python
input_dropdown = alt.binding_select(options=['Grande Niterói', 'Capital', 'Baixada Fluminense', 'Interior'], name='Região')
selection = alt.selection_point(fields=['regiao'], bind=input_dropdown)

cor = (
    alt.when(selection)
    .then(alt.Color("regiao:N").legend(None))
    .otherwise(alt.value("lightgray"))
)

alt.Chart(dfanos).mark_circle().encode(
    alt.X('quantidade_registro_posse_drogas', axis=alt.Axis(title='Quantidade de registros de posse de drogas')),
    alt.Y('quantidade_registro_trafico_drogas', axis=alt.Axis(title='Quantidade de registros de tráfico')),
    alt.Size('count()'),
    color= cor
).properties(
    width= 1000,
    title='Relação entre tráfico e posse de drogas'
).add_params(
    selection
).interactive()
```

Me surpreendi com o fato de que muitos lugares tem bastante registro de posse mas pouco de tráfico. Talvez isso ocorra, pois o tráfico é menos denunciado e mais difícil de encontrar e combater.
